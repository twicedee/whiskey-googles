```python
def clean_data(df):
    """Clean the dataset and update image paths"""
    # 1. Fill missing values
    df["proof"] = df["proof"].fillna(0)      # Missing proof -> 0
    df["size"] = df["size"].fillna(750)      # Missing size -> 750mL
    
    # 2. Update image_url to local paths (e.g., "images/10_0.jpg")
    df["image_url"] = df.apply(
        lambda row: f"images/{int(row['id'])}_{row.name}.jpg" 
        if pd.notna(row["id"]) else None,
        axis=1
    )
    
    return df

df = clean_data(data)

df.to_csv("dataset/processed/clean_image_whisky_dataset.csv", index=False)
```
